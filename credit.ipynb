```python
import numpy as np
#  Define your direct conditions
conditions = [
    (df['Customer_Age'] >= 20) & (df['Customer_Age'] <= 30),
    (df['Customer_Age'] > 30) & (df['Customer_Age'] <= 40),
    (df['Customer_Age'] > 40) & (df['Customer_Age'] <= 50),
    (df['Customer_Age'] > 50) & (df['Customer_Age'] <= 60),
    (df['Customer_Age'] > 60)
]

#  Match them to your exact labels
labels = ['20-30', '30-40', '40-50', '50-60', '60+']

#  Apply directly to the column
df['Age_Group'] = np.select(conditions, labels, default='Unknown')
```

```python
df
```

```python

conditions = [
    df['Income'] < 35000,                               # Less than or equal to 35k
    (df['Income'] >= 35000) & (df['Income'] < 70000),     # Between 35k and 70k
    df['Income'] >= 70000                                # Greater than or equal to 70k
]

# 2. Match them to your exact labels
labels = ['Low', 'Medium', 'High']

# 3. Create the new column instantly for the whole dataset
df['Income_Group'] = np.select(conditions, labels, default='Unknown')
```

```python
from sqlalchemy import create_engine

#MYSQL connection
username='root'
password='123456'
host='localhost'
port='3306'
database='credit'

engine= create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")

#write dataframe to MYSQL
table_name="customer_credit"
df.to_sql(table_name, engine ,if_exists='replace', index=False)

#Read back sample
pd.read_sql("SELECT * FROM customer_credit LIMIT 5;",engine)
```
